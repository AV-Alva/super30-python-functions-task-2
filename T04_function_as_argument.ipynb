```python
def add(a, b):
    """Return addition of two numbers."""
    return a + b


def multiply(a, b):
    """Return multiplication of two numbers."""
    return a * b
```

```python
def calculate(operation, a, b):
    """Execute the supplied mathematical function."""
    return operation(a, b)
```

```python
print("Addition Operation:", calculate(add, 10, 20))

print("Multiplication Operation:", calculate(multiply, 10, 20))
```
