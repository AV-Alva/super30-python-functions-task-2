```python
def find_largest(*numbers):
    """Return the largest supplied number."""
    if len(numbers) == 0:
        return None

    largest = numbers[0]

    for i in numbers:
        if i > largest:
            largest = i

    return largest


print("Largest of 10, 40, 25, 80, 30:", find_largest(10, 40, 25, 80, 30))
```
