```python
def create_profile(**kwargs):
    """Print dynamically supplied user information."""
    for key, value in kwargs.items():
        print(key, ":", value)
        
create_profile(name="Rahul", age=22, course="Python", city="Bangalore")
create_profile(name="Anitha", age=21, course="Gen AI", city="Mysore", country="India")
```
